```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV files
phi_df = pd.read_csv('./CSV Files/lstm_training_summary.csv')  # Columns: 'epoch', 'phi'
shap_df = pd.read_csv('../LSTM_SHAP_VALUE/lstm_shap_results_20260226_094442.csv')  # Columns: 'epoch', 'shap_total'

# Add epoch if missing
if 'epoch' not in phi_df.columns:
    phi_df['epoch'] = range(1, len(phi_df) + 1)
if 'epoch' not in shap_df.columns:
    shap_df['epoch'] = range(1, len(shap_df) + 1)

# Merge on epoch
merged_df = pd.merge(phi_df, shap_df, on='epoch')

# Compute correlations
pearson_corr, pearson_p = pearsonr(merged_df['phi'], merged_df['shap_total'])
spearman_corr, spearman_p = spearmanr(merged_df['phi'], merged_df['shap_total'])

print(f"Pearson correlation: {pearson_corr:.4f}, p-value: {pearson_p:.4e}")
print(f"Spearman correlation: {spearman_corr:.4f}, p-value: {spearman_p:.4e}")

# Plot
plt.figure(figsize=(8,6))
sns.regplot(x='phi', y='shap_total', data=merged_df, ci=None,
            scatter_kws={'s':50, 'alpha':0.6}, line_kws={'color':'red'})
plt.title('Correlation between Φ* and SHAP values (LSTM)', fontsize=14)
plt.xlabel('Φ* (phi)', fontsize=12)
plt.ylabel('SHAP Total', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Annotate correlation coefficients
plt.text(0.05, 0.95, f"Pearson r = {pearson_corr:.4f}\nSpearman ρ = {spearman_corr:.4f}",
         transform=plt.gca().transAxes,
         fontsize=12, verticalalignment='top',
         bbox=dict(facecolor='white', alpha=0.6, edgecolor='gray'))

plt.tight_layout()


import os
SAVE_DIR = "./SHAP_PNG Files(TimesNewRoman)"
os.makedirs(SAVE_DIR, exist_ok=True)
plt.savefig(f"{SAVE_DIR}/Correlation between Φ* and SHAP values (LSTM)", dpi=300)
plt.show()
```

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Function to load and prepare data
def load_data(phi_path, shap_path):
    phi_df = pd.read_csv(phi_path)
    shap_df = pd.read_csv(shap_path)

    if 'epoch' not in phi_df.columns:
        phi_df['epoch'] = range(1, len(phi_df) + 1)
    if 'epoch' not in shap_df.columns:
        shap_df['epoch'] = range(1, len(shap_df) + 1)

    merged_df = pd.merge(phi_df, shap_df, on='epoch')
    return merged_df

# Load datasets
lstm_df = load_data('./CSV Files/lstm_training_summary.csv',
                    '../LSTM_SHAP_VALUE/lstm_shap_results_20260226_094442.csv')

rnn_df = load_data('./CSV Files/rnn_training_summary.csv',
                   '../RNN_SHAP_VALUE/rnn_shap_results_final.csv')

gru_df = load_data('./CSV Files/gru_training_summary.csv',
                   '../GRU_SHAP_VALUE/gru_shap_results_20260226_092929.csv')

# Plot (scaled figure)
plt.figure(figsize=(10,8))

# LSTM
sns.regplot(x='phi', y='shap_total', data=lstm_df, ci=None,
            scatter_kws={'s':50, 'alpha':0.6},
            line_kws={'color':'blue'},
            label='LSTM')

# RNN
sns.regplot(x='phi', y='shap_total', data=rnn_df, ci=None,
            scatter_kws={'s':50, 'alpha':0.6},
            line_kws={'color':'green'},
            label='RNN')

# GRU
sns.regplot(x='phi', y='shap_total', data=gru_df, ci=None,
            scatter_kws={'s':50, 'alpha':0.6},
            line_kws={'color':'red'},
            label='GRU')

# Axis scaling
plt.ylim(0, 0.3)

plt.title('Correlation between Φ* and SHAP values (All Models)', fontsize=14)
plt.xlabel('Φ* (phi)', fontsize=12)
plt.ylabel('SHAP Total', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()

# Save
SAVE_DIR = "./SHAP_PNG Files(TimesNewRoman)"
os.makedirs(SAVE_DIR, exist_ok=True)
plt.savefig(f"{SAVE_DIR}/Combined_Correlation_All_Models_scaled.png", dpi=300)

plt.show()
```
