# GRU (4 units) + Monte-Carlo TPM + Big Φ (SIA) 

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import itertools
import matplotlib.pyplot as plt
import seaborn as sns
import pyphi

# ---------------------------------------------------------------
# PyPhi configuration
# ---------------------------------------------------------------
pyphi.config.VALIDATE_SUBSYSTEM_STATES = False
pyphi.config.VALIDATE_CONDITIONAL_INDEPENDENCE = False

device = torch.device("cpu")

# ---------------------------------------------------------------
# PART 1 — Parity Dataset
# ---------------------------------------------------------------
def generate_parity_data(n_samples=2000):
    X = np.random.randint(0, 2, (n_samples, 2))
    y = np.sum(X, axis=1) % 2
    return X.astype(np.float32), y.astype(np.float32)

X, y = generate_parity_data(2000)
X_train, y_train = X[:1600], y[:1600]
X_test, y_test = X[1600:], y[1600:]

train_loader = DataLoader(
    TensorDataset(torch.tensor(X_train), torch.tensor(y_train)),
    batch_size=64,
    shuffle=True
)

# ---------------------------------------------------------------
# PART 2 — GRU Model
# ---------------------------------------------------------------
input_size = 2
hidden_size = 4

class GRURegressor(nn.Module):
    def __init__(self):
        super().__init__()
        self.gru = nn.GRU(input_size, hidden_size, batch_first=True)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x, h0=None):
        x = x.unsqueeze(1)
        _, h = self.gru(x, h0)
        out = self.fc(h.squeeze(0))
        return torch.sigmoid(out), h

model = GRURegressor().to(device)
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# ---------------------------------------------------------------
# PART 3 — Hidden State Space
# ---------------------------------------------------------------
n_nodes = hidden_size
states = list(itertools.product([0, 1], repeat=n_nodes))
num_states = len(states)

mc_inputs_per_state = 400
mc_bernoulli_samples = 4
rng = np.random.RandomState(42)

connectivity = np.ones((n_nodes, n_nodes), dtype=int)

def idx_to_state(idx):
    return np.array(list(np.binary_repr(idx, width=n_nodes))).astype(int)

# ---------------------------------------------------------------
# PART 4 — Compute TPM + Big Φ (SIA)
# ---------------------------------------------------------------
def compute_big_phi_and_tpm(model):
    tpm_counts = np.zeros((num_states, num_states))

    with torch.no_grad():
        for i, h_state in enumerate(states):
            h0 = torch.tensor(h_state, dtype=torch.float32).view(1,1,-1)
            for _ in range(mc_inputs_per_state):
                x = rng.randint(0,2,size=(input_size,)).astype(np.float32)
                x_tensor = torch.tensor(x).view(1,1,-1)
                _, h_next = model.gru(x_tensor, h0)
                probs = torch.sigmoid(h_next).squeeze().numpy()

                for _ in range(mc_bernoulli_samples):
                    s = (rng.rand(n_nodes) < probs).astype(int)
                    j = int("".join(map(str,s)),2)
                    tpm_counts[i,j] += 1

    tpm = tpm_counts + 1e-8
    tpm /= tpm.sum(axis=1, keepdims=True)

    network = pyphi.Network(tpm, connectivity)

    phi_vals, sia_objects = [], []
    for i in range(num_states):
        st = idx_to_state(i)
        try:
            subsystem = pyphi.Subsystem(network, st, tuple(range(n_nodes)))
            sia = pyphi.compute.sia(subsystem)
            phi_vals.append(sia.phi)
            sia_objects.append(sia)
        except:
            phi_vals.append(0.0)
            sia_objects.append(None)

    best_idx = np.argmax(phi_vals)
    return phi_vals[best_idx], tpm, sia_objects[best_idx], network

# ---------------------------------------------------------------
# PART 5 — Accuracy computation
# ---------------------------------------------------------------
def compute_accuracy(model):
    model.eval()
    with torch.no_grad():
        x = torch.tensor(X_test)
        y_true = torch.tensor(y_test)
        preds,_ = model(x)
        preds = (preds.squeeze() > 0.5).float()
        return (preds == y_true).float().mean().item()

# ---------------------------------------------------------------
# PART 6 — Training loop
# ---------------------------------------------------------------
EPOCHS = 10
phi_list, acc_list = [], []

print("\nTRAINING GRU + IIT ANALYSIS\n")

for ep in range(1, EPOCHS+1):
    model.train()
    for xb,yb in train_loader:
        optimizer.zero_grad()
        pred,_ = model(xb)
        loss = criterion(pred.squeeze(), yb)
        loss.backward()
        optimizer.step()

    acc = compute_accuracy(model)
    phi, tpm, best_sia, network = compute_big_phi_and_tpm(model)

    phi_list.append(phi)
    acc_list.append(acc)

    print(f"Epoch {ep}: Accuracy={acc:.3f} | Big Φ={phi:.6f}")

# ---------------------------------------------------------------
# PART 7 — Big Φ vs Epoch
# ---------------------------------------------------------------
plt.figure(figsize=(7,5))
plt.plot(range(1,EPOCHS+1), phi_list, marker='o')
plt.xlabel("Epoch")
plt.ylabel("Big Φ (SIA)")
plt.title("Big Φ vs Epoch")
plt.grid(True)
plt.show()

# ---------------------------------------------------------------
# PART 8 — Big Φ vs Accuracy
# ---------------------------------------------------------------
plt.figure(figsize=(7,5))
plt.plot(acc_list, phi_list, marker='o')
plt.xlabel("Accuracy")
plt.ylabel("Big Φ (SIA)")
plt.title("Big Φ vs Accuracy")
plt.grid(True)
plt.show()

# ---------------------------------------------------------------
# PART 9 — MIP-Cut TPM Visualization (MNIST-style)
# ---------------------------------------------------------------
subsystem = pyphi.Subsystem(
    network,
    idx_to_state(np.argmax(phi_list)),
    tuple(range(n_nodes))
)

sia = pyphi.compute.sia(subsystem)
cut = sia.cut

cut_subsystem = subsystem.apply_cut(cut)
cut_tpm_sbn = cut_subsystem.tpm
cut_tpm = pyphi.convert.state_by_node2state_by_state(cut_tpm_sbn)

diff = tpm - cut_tpm

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
sns.heatmap(tpm, cmap="viridis")
plt.title("Original TPM")

plt.subplot(1,3,2)
sns.heatmap(cut_tpm, cmap="viridis")
plt.title(f"MIP-Cut TPM (Φ={sia.phi:.2f})")

plt.subplot(1,3,3)
sns.heatmap(diff, cmap="coolwarm", center=0)
plt.title("TPM Difference")

plt.tight_layout()
plt.show()
```

# GRU (4 units) + Big Φ (IIT-SIA) + Φ* (Barrett & Seth)

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import itertools
import matplotlib.pyplot as plt
import pyphi
from itertools import combinations
from scipy.stats import pearsonr

# ---------------------------------------------------------------
# PyPhi config
# ---------------------------------------------------------------
pyphi.config.VALIDATE_SUBSYSTEM_STATES = False
pyphi.config.VALIDATE_CONDITIONAL_INDEPENDENCE = False

device = torch.device("cpu")

# ---------------------------------------------------------------
# PART 1 — Parity Dataset
# ---------------------------------------------------------------
def generate_parity_data(n_samples=2000):
    X = np.random.randint(0, 2, (n_samples, 2))
    y = np.sum(X, axis=1) % 2
    return X.astype(np.float32), y.astype(np.float32)

X, y = generate_parity_data()
X_train, y_train = X[:1600], y[:1600]
X_test, y_test = X[1600:], y[1600:]

train_loader = DataLoader(
    TensorDataset(torch.tensor(X_train), torch.tensor(y_train)),
    batch_size=64,
    shuffle=True
)

# ---------------------------------------------------------------
# PART 2 — GRU Model (4 hidden units)
# ---------------------------------------------------------------
input_size = 2
hidden_size = 4

class GRURegressor(nn.Module):
    def __init__(self):
        super().__init__()
        self.gru = nn.GRU(input_size, hidden_size, batch_first=True)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x, h0=None):
        x = x.unsqueeze(1)
        _, h = self.gru(x, h0)
        out = self.fc(h.squeeze(0))
        return torch.sigmoid(out), h

model = GRURegressor()
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# ---------------------------------------------------------------
# PART 3 — State space & Monte-Carlo params
# ---------------------------------------------------------------
n_nodes = hidden_size
states = list(itertools.product([0,1], repeat=n_nodes))
num_states = len(states)

mc_inputs_per_state = 400
mc_bernoulli_samples = 4
rng = np.random.RandomState(42)

connectivity = np.ones((n_nodes, n_nodes), dtype=int)

def idx_to_state(idx):
    return np.array(list(np.binary_repr(idx, width=n_nodes))).astype(int)

# ---------------------------------------------------------------
# PART 4 — Compute Big Φ (IIT / SIA)
# ---------------------------------------------------------------
def compute_big_phi_and_tpm(model):
    tpm_counts = np.zeros((num_states, num_states))

    with torch.no_grad():
        for i, h_state in enumerate(states):
            h0 = torch.tensor(h_state, dtype=torch.float32).view(1,1,-1)

            for _ in range(mc_inputs_per_state):
                x = rng.randint(0,2,size=(input_size,)).astype(np.float32)
                x_tensor = torch.tensor(x).view(1,1,-1)

                _, h_next = model.gru(x_tensor, h0)
                probs = torch.sigmoid(h_next).squeeze().numpy()

                for _ in range(mc_bernoulli_samples):
                    s = (rng.rand(n_nodes) < probs).astype(int)
                    j = int("".join(map(str,s)),2)
                    tpm_counts[i,j] += 1

    tpm = tpm_counts + 1e-8
    tpm /= tpm.sum(axis=1, keepdims=True)

    network = pyphi.Network(tpm, connectivity)

    phi_vals = []
    for i in range(num_states):
        try:
            subsystem = pyphi.Subsystem(network, idx_to_state(i), tuple(range(n_nodes)))
            sia = pyphi.compute.sia(subsystem)
            phi_vals.append(sia.phi)
        except:
            phi_vals.append(0.0)

    return max(phi_vals), tpm

# ---------------------------------------------------------------
# PART 5 — Φ* (Barrett & Seth)
# ---------------------------------------------------------------
def mutual_information(joint):
    px = joint.sum(axis=1, keepdims=True)
    py = joint.sum(axis=0, keepdims=True)
    eps = 1e-12
    joint += eps; px += eps; py += eps
    return np.sum(joint * np.log2(joint / (px @ py)))

def compute_phi_star(tpm, n_nodes):
    p_xt = np.ones(tpm.shape[0]) / tpm.shape[0]
    joint = p_xt[:,None] * tpm
    I_whole = mutual_information(joint)

    states = list(itertools.product([0,1], repeat=n_nodes))
    min_phi_star = np.inf
    nodes = list(range(n_nodes))

    for k in range(1, n_nodes//2 + 1):
        for part in combinations(nodes, k):
            comp = tuple(set(nodes) - set(part))

            def project(subset):
                mapping = {}
                for i,s in enumerate(states):
                    key = tuple(s[j] for j in subset)
                    mapping.setdefault(key, []).append(i)
                return mapping

            I_cut = 0.0
            for group in [project(part), project(comp)]:
                joint_sub = np.zeros((len(group), len(group)))
                keys = list(group.keys())
                for i,ki in enumerate(keys):
                    for j,kj in enumerate(keys):
                        for si in group[ki]:
                            for sj in group[kj]:
                                joint_sub[i,j] += joint[si,sj]
                I_cut += mutual_information(joint_sub)

            min_phi_star = min(min_phi_star, I_whole - I_cut)

    return max(min_phi_star, 0.0)

# ---------------------------------------------------------------
# PART 6 — Accuracy
# ---------------------------------------------------------------
def compute_accuracy(model):
    with torch.no_grad():
        preds,_ = model(torch.tensor(X_test))
        preds = (preds.squeeze() > 0.5).float()
        return (preds == torch.tensor(y_test)).float().mean().item()

# ---------------------------------------------------------------
# PART 7 — Training + Φ & Φ*
# ---------------------------------------------------------------
EPOCHS = 10
phi_list, phi_star_list, acc_list = [], [], []

print("\nTRAINING GRU + Φ and Φ*\n")

for ep in range(1, EPOCHS+1):
    model.train()
    for xb,yb in train_loader:
        optimizer.zero_grad()
        pred,_ = model(xb)
        loss = criterion(pred.squeeze(), yb)
        loss.backward()
        optimizer.step()

    acc = compute_accuracy(model)
    phi, tpm = compute_big_phi_and_tpm(model)
    phi_star = compute_phi_star(tpm, n_nodes)

    phi_list.append(phi)
    phi_star_list.append(phi_star)
    acc_list.append(acc)

    print(f"Epoch {ep} | Acc={acc:.3f} | Φ={phi:.6f} | Φ*={phi_star:.6f}")

# ---------------------------------------------------------------
# PART 8 — Plots
# ---------------------------------------------------------------
plt.figure(figsize=(7,5))
plt.plot(phi_list, label="Big Φ (IIT)", marker='o')
plt.plot(phi_star_list, label="Φ* (Barrett)", marker='s')
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.title("Φ and Φ* vs Epoch")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,5))
plt.scatter(phi_list, phi_star_list)
plt.xlabel("Big Φ (IIT)")
plt.ylabel("Φ* (Barrett)")
plt.title("Φ vs Φ* Correlation")
plt.grid(True)
plt.tight_layout()
plt.show()

r, p = pearsonr(phi_list, phi_star_list)
print(f"Correlation: r = {r:.3f}, p = {p:.4f}")
```

# Comparison of PHI VS PHI STAR of BARETT AND SETH VS PHI STAR OF OZUMI(DECODING PERSPECTIVE)

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import itertools
import matplotlib.pyplot as plt
import pyphi
from itertools import combinations
from scipy.stats import pearsonr

# ---------------------------------------------------------------
# PyPhi config
# ---------------------------------------------------------------
pyphi.config.VALIDATE_SUBSYSTEM_STATES = False
pyphi.config.VALIDATE_CONDITIONAL_INDEPENDENCE = False

device = torch.device("cpu")

# ---------------------------------------------------------------
# PART 1 — Parity Dataset
# ---------------------------------------------------------------
def generate_parity_data(n_samples=2000):
    X = np.random.randint(0, 2, (n_samples, 2))
    y = np.sum(X, axis=1) % 2
    return X.astype(np.float32), y.astype(np.float32)

X, y = generate_parity_data()
X_train, y_train = X[:1600], y[:1600]
X_test, y_test = X[1600:], y[1600:]

train_loader = DataLoader(
    TensorDataset(torch.tensor(X_train), torch.tensor(y_train)),
    batch_size=64,
    shuffle=True
)

# ---------------------------------------------------------------
# PART 2 — GRU Model (configurable hidden units)
# ---------------------------------------------------------------
input_size = 2
hidden_size = 4  # Start with 4, can scale to 8, 16, 32, etc.

class GRURegressor(nn.Module):
    def __init__(self, hidden_dim):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.gru = nn.GRU(input_size, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x, h0=None):
        x = x.unsqueeze(1)
        _, h = self.gru(x, h0)
        out = self.fc(h.squeeze(0))
        return torch.sigmoid(out), h

model = GRURegressor(hidden_size)
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# ---------------------------------------------------------------
# PART 3 — State space & Monte-Carlo params
# ---------------------------------------------------------------
n_nodes = hidden_size
states = list(itertools.product([0,1], repeat=n_nodes))
num_states = len(states)

mc_inputs_per_state = 400
mc_bernoulli_samples = 4
rng = np.random.RandomState(42)

connectivity = np.ones((n_nodes, n_nodes), dtype=int)

def idx_to_state(idx):
    return np.array(list(np.binary_repr(idx, width=n_nodes))).astype(int)

# ---------------------------------------------------------------
# PART 4 — Compute Big Φ (IIT / SIA) - Only for small models
# ---------------------------------------------------------------
def compute_big_phi_and_tpm(model):
    """Only feasible for hidden_size <= 6"""
    if hidden_size > 6:
        print("⚠️  Skipping full IIT Φ computation (too large)")
        return None, None
    
    tpm_counts = np.zeros((num_states, num_states))

    with torch.no_grad():
        for i, h_state in enumerate(states):
            h0 = torch.tensor(h_state, dtype=torch.float32).view(1,1,-1)

            for _ in range(mc_inputs_per_state):
                x = rng.randint(0,2,size=(input_size,)).astype(np.float32)
                x_tensor = torch.tensor(x).view(1,1,-1)

                _, h_next = model.gru(x_tensor, h0)
                probs = torch.sigmoid(h_next).squeeze().numpy()

                for _ in range(mc_bernoulli_samples):
                    s = (rng.rand(n_nodes) < probs).astype(int)
                    j = int("".join(map(str,s)),2)
                    tpm_counts[i,j] += 1

    tpm = tpm_counts + 1e-8
    tpm /= tpm.sum(axis=1, keepdims=True)

    network = pyphi.Network(tpm, connectivity)

    phi_vals = []
    for i in range(num_states):
        try:
            subsystem = pyphi.Subsystem(network, idx_to_state(i), tuple(range(n_nodes)))
            sia = pyphi.compute.sia(subsystem)
            phi_vals.append(sia.phi)
        except:
            phi_vals.append(0.0)

    return max(phi_vals), tpm

# ---------------------------------------------------------------
# PART 5 — Φ* (Barrett & Seth) - Original version
# ---------------------------------------------------------------
def mutual_information(joint):
    px = joint.sum(axis=1, keepdims=True)
    py = joint.sum(axis=0, keepdims=True)
    eps = 1e-12
    joint += eps; px += eps; py += eps
    return np.sum(joint * np.log2(joint / (px @ py)))

def compute_phi_star_barrett(tpm, n_nodes):
    """Original Barrett & Seth Φ* using MI decomposition"""
    if tpm is None:
        return None
    
    p_xt = np.ones(tpm.shape[0]) / tpm.shape[0]
    joint = p_xt[:,None] * tpm
    I_whole = mutual_information(joint)

    states = list(itertools.product([0,1], repeat=n_nodes))
    min_phi_star = np.inf
    nodes = list(range(n_nodes))

    for k in range(1, n_nodes//2 + 1):
        for part in combinations(nodes, k):
            comp = tuple(set(nodes) - set(part))

            def project(subset):
                mapping = {}
                for i,s in enumerate(states):
                    key = tuple(s[j] for j in subset)
                    mapping.setdefault(key, []).append(i)
                return mapping

            I_cut = 0.0
            for group in [project(part), project(comp)]:
                joint_sub = np.zeros((len(group), len(group)))
                keys = list(group.keys())
                for i,ki in enumerate(keys):
                    for j,kj in enumerate(keys):
                        for si in group[ki]:
                            for sj in group[kj]:
                                joint_sub[i,j] += joint[si,sj]
                I_cut += mutual_information(joint_sub)

            min_phi_star = min(min_phi_star, I_whole - I_cut)

    return max(min_phi_star, 0.0)

# ---------------------------------------------------------------
# PART 6 — Decoder-based Φ* with Gaussian Approximation (SCALES!)
# ---------------------------------------------------------------
def collect_hidden_states(model, n_samples=2000):
    """Collect hidden state trajectories from the GRU"""
    states_list = []
    
    with torch.no_grad():
        h = None
        for _ in range(n_samples):
            x = rng.randint(0, 2, size=(1, input_size)).astype(np.float32)
            x_tensor = torch.tensor(x).view(1, 1, -1)
            
            _, h = model.gru(x_tensor, h)
            states_list.append(h.squeeze().numpy())
    
    return np.array(states_list)

def compute_phi_star_decoder_gaussian(model, n_samples=2000):
    """
    Decoder-based Φ* using Gaussian approximation.
    Measures information loss when splitting the system.
    
    This scales to large models! O(n³) instead of O(2^n)
    """
    # Collect state trajectories
    states = collect_hidden_states(model, n_samples)
    n_nodes = states.shape[1]
    
    # Compute covariance matrices for Gaussian approximation
    # States at time t and t-1
    X_t = states[1:]  # Current states
    X_tm1 = states[:-1]  # Previous states
    
    # Center the data
    X_t_mean = X_t.mean(axis=0)
    X_tm1_mean = X_tm1.mean(axis=0)
    X_t_centered = X_t - X_t_mean
    X_tm1_centered = X_tm1 - X_tm1_mean
    
    # Compute covariances
    Cov_t = np.cov(X_t_centered.T) + 1e-6 * np.eye(n_nodes)
    Cov_tm1 = np.cov(X_tm1_centered.T) + 1e-6 * np.eye(n_nodes)
    Cov_joint = np.cov(np.hstack([X_tm1_centered, X_t_centered]).T)
    
    # Total mutual information I(X_t; X_{t-1})
    # For Gaussian: MI = 0.5 * log(|Cov_t| * |Cov_tm1| / |Cov_joint|)
    sign_t, logdet_t = np.linalg.slogdet(Cov_t)
    sign_tm1, logdet_tm1 = np.linalg.slogdet(Cov_tm1)
    sign_joint, logdet_joint = np.linalg.slogdet(Cov_joint)
    
    I_whole = 0.5 * (logdet_t + logdet_tm1 - logdet_joint) / np.log(2)  # Convert to bits
    
    # Find minimum information partition (MIP)
    nodes = list(range(n_nodes))
    min_phi_decoder = np.inf
    
    for k in range(1, n_nodes // 2 + 1):
        for partition in combinations(nodes, k):
            part_A = list(partition)
            part_B = list(set(nodes) - set(partition))
            
            # Compute MI for each partition independently
            I_cut = 0.0
            
            for part in [part_A, part_B]:
                if len(part) == 0:
                    continue
                
                # Extract submatrices for this partition
                Cov_t_sub = Cov_t[np.ix_(part, part)]
                Cov_tm1_sub = Cov_tm1[np.ix_(part, part)]
                
                # Joint covariance for this partition
                part_indices = part + [p + n_nodes for p in part]
                Cov_joint_sub = Cov_joint[np.ix_(part_indices, part_indices)]
                
                # MI for this partition
                sign_t_sub, logdet_t_sub = np.linalg.slogdet(Cov_t_sub)
                sign_tm1_sub, logdet_tm1_sub = np.linalg.slogdet(Cov_tm1_sub)
                sign_joint_sub, logdet_joint_sub = np.linalg.slogdet(Cov_joint_sub)
                
                I_part = 0.5 * (logdet_t_sub + logdet_tm1_sub - logdet_joint_sub) / np.log(2)
                I_cut += I_part
            
            # Φ for this partition
            phi_partition = I_whole - I_cut
            min_phi_decoder = min(min_phi_decoder, phi_partition)
    
    return max(min_phi_decoder, 0.0)

# ---------------------------------------------------------------
# PART 7 — Φ Approximation for Large Models (Tononi et al. 2016)
# ---------------------------------------------------------------
def compute_phi_approximation(model, n_samples=2000):
    """
    Fast Φ approximation based on Tononi et al. (2016)
    Uses singular value decomposition of the interaction matrix.
    
    Scales to large models! O(n³)
    """
    # Collect state trajectories
    states = collect_hidden_states(model, n_samples)
    n_nodes = states.shape[1]
    
    X_t = states[1:]
    X_tm1 = states[:-1]
    
    # Center data
    X_t_centered = X_t - X_t.mean(axis=0)
    X_tm1_centered = X_tm1 - X_tm1.mean(axis=0)
    
    # Estimate interaction matrix via linear regression
    # X_t ≈ W @ X_{t-1}
    W = np.linalg.lstsq(X_tm1_centered, X_t_centered, rcond=None)[0].T
    
    # Compute covariance matrices
    Cov_t = np.cov(X_t_centered.T) + 1e-6 * np.eye(n_nodes)
    Cov_tm1 = np.cov(X_tm1_centered.T) + 1e-6 * np.eye(n_nodes)
    
    # SVD-based approximation
    U, S, Vt = np.linalg.svd(W)
    
    # Φ approximation using effective information
    # Sum of squared singular values weighted by variance
    var_tm1 = np.diag(Cov_tm1)
    phi_approx = 0.0
    
    for i in range(min(n_nodes, len(S))):
        # Effective information carried by each singular vector
        ei = S[i]**2 * var_tm1[i] if i < len(var_tm1) else S[i]**2
        phi_approx += ei
    
    # Normalize by system size for comparability
    phi_approx = phi_approx / n_nodes
    
    return phi_approx

# ---------------------------------------------------------------
# PART 8 — Accuracy
# ---------------------------------------------------------------
def compute_accuracy(model):
    with torch.no_grad():
        preds,_ = model(torch.tensor(X_test))
        preds = (preds.squeeze() > 0.5).float()
        return (preds == torch.tensor(y_test)).float().mean().item()

# ---------------------------------------------------------------
# PART 9 — Training + All Φ measures
# ---------------------------------------------------------------
EPOCHS = 10
phi_list, phi_star_barrett_list, phi_star_decoder_list, phi_approx_list, acc_list = [], [], [], [], []

print(f"\n{'='*60}")
print(f"TRAINING GRU (hidden_size={hidden_size})")
print(f"{'='*60}\n")

for ep in range(1, EPOCHS+1):
    model.train()
    for xb,yb in train_loader:
        optimizer.zero_grad()
        pred,_ = model(xb)
        loss = criterion(pred.squeeze(), yb)
        loss.backward()
        optimizer.step()

    acc = compute_accuracy(model)
    
    # Full IIT Φ (only for small models)
    if hidden_size <= 6:
        phi, tpm = compute_big_phi_and_tpm(model)
        phi_star_barrett = compute_phi_star_barrett(tpm, n_nodes)
    else:
        phi = None
        phi_star_barrett = None
    
    # Scalable measures (work for any size!)
    phi_star_decoder = compute_phi_star_decoder_gaussian(model)
    phi_approx = compute_phi_approximation(model)
    
    phi_list.append(phi)
    phi_star_barrett_list.append(phi_star_barrett)
    phi_star_decoder_list.append(phi_star_decoder)
    phi_approx_list.append(phi_approx)
    acc_list.append(acc)

    if phi is not None:
        print(f"Epoch {ep} | Acc={acc:.3f} | Φ_IIT={phi:.6f} | Φ*_Barrett={phi_star_barrett:.6f} | Φ*_Decoder={phi_star_decoder:.6f} | Φ_Approx={phi_approx:.6f}")
    else:
        print(f"Epoch {ep} | Acc={acc:.3f} | Φ*_Decoder={phi_star_decoder:.6f} | Φ_Approx={phi_approx:.6f}")

# ---------------------------------------------------------------
# PART 10 — Plots & Correlations
# ---------------------------------------------------------------
print(f"\n{'='*60}")
print("CORRELATIONS")
print(f"{'='*60}")

if hidden_size <= 6:
    # Plot all measures
    plt.figure(figsize=(10,6))
    plt.plot(phi_list, label="Φ (IIT)", marker='o')
    plt.plot(phi_star_barrett_list, label="Φ* (Barrett)", marker='s')
    plt.plot(phi_star_decoder_list, label="Φ* (Decoder)", marker='^')
    plt.plot(phi_approx_list, label="Φ (Approximation)", marker='d')
    plt.xlabel("Epoch")
    plt.ylabel("Value")
    plt.title(f"Integration Measures vs Epoch (hidden_size={hidden_size})")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Correlation matrix
    measures = {
        'Φ_IIT': phi_list,
        'Φ*_Barrett': phi_star_barrett_list,
        'Φ*_Decoder': phi_star_decoder_list,
        'Φ_Approx': phi_approx_list
    }
    
    print("\nPairwise correlations:")
    for m1 in measures:
        for m2 in measures:
            if m1 < m2:
                r, p = pearsonr(measures[m1], measures[m2])
                print(f"  {m1} vs {m2}: r={r:.3f}, p={p:.4f}")
    
    # Scatter plots
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    
    axes[0,0].scatter(phi_list, phi_star_barrett_list)
    axes[0,0].set_xlabel("Φ (IIT)")
    axes[0,0].set_ylabel("Φ* (Barrett)")
    axes[0,0].set_title("IIT vs Barrett")
    axes[0,0].grid(True)
    
    axes[0,1].scatter(phi_list, phi_star_decoder_list)
    axes[0,1].set_xlabel("Φ (IIT)")
    axes[0,1].set_ylabel("Φ* (Decoder)")
    axes[0,1].set_title("IIT vs Decoder")
    axes[0,1].grid(True)
    
    axes[1,0].scatter(phi_star_barrett_list, phi_star_decoder_list)
    axes[1,0].set_xlabel("Φ* (Barrett)")
    axes[1,0].set_ylabel("Φ* (Decoder)")
    axes[1,0].set_title("Barrett vs Decoder")
    axes[1,0].grid(True)
    
    axes[1,1].scatter(phi_approx_list, phi_star_decoder_list)
    axes[1,1].set_xlabel("Φ (Approximation)")
    axes[1,1].set_ylabel("Φ* (Decoder)")
    axes[1,1].set_title("Approximation vs Decoder")
    axes[1,1].grid(True)
    
    plt.tight_layout()
    plt.show()
    
else:
    # For large models, only plot scalable measures
    plt.figure(figsize=(8,5))
    plt.plot(phi_star_decoder_list, label="Φ* (Decoder)", marker='^')
    plt.plot(phi_approx_list, label="Φ (Approximation)", marker='d')
    plt.xlabel("Epoch")
    plt.ylabel("Value")
    plt.title(f"Scalable Integration Measures (hidden_size={hidden_size})")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    r, p = pearsonr(phi_star_decoder_list, phi_approx_list)
    print(f"\nΦ_Approx vs Φ*_Decoder: r={r:.3f}, p={p:.4f}")

print(f"\n{'='*60}")
print("SCALING TO LARGER MODELS")
print(f"{'='*60}")
print(f"""
To scale up to larger models (e.g., hidden_size=16, 32, 64):
1. Change 'hidden_size' at the top of the script
2. Full IIT Φ will be skipped automatically (only feasible for ≤6 units)
3. Φ*_Decoder and Φ_Approx will still compute efficiently!

These scalable measures use O(n³) operations instead of O(2^n),
making them practical for networks with hundreds of units.
""")
```

# Comparison of the phi star (newer version of barett and seth approach)

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import itertools
import matplotlib.pyplot as plt
import pyphi
from itertools import combinations
from scipy.stats import pearsonr

# ---------------------------------------------------------------
# PyPhi config
# ---------------------------------------------------------------
pyphi.config.VALIDATE_SUBSYSTEM_STATES = False
pyphi.config.VALIDATE_CONDITIONAL_INDEPENDENCE = False

device = torch.device("cpu")

# ---------------------------------------------------------------
# PART 1 — Parity Dataset
# ---------------------------------------------------------------
def generate_parity_data(n_samples=2000):
    X = np.random.randint(0, 2, (n_samples, 2))
    y = np.sum(X, axis=1) % 2
    return X.astype(np.float32), y.astype(np.float32)

X, y = generate_parity_data()
X_train, y_train = X[:1600], y[:1600]
X_test, y_test = X[1600:], y[1600:]

train_loader = DataLoader(
    TensorDataset(torch.tensor(X_train), torch.tensor(y_train)),
    batch_size=64,
    shuffle=True
)

# ---------------------------------------------------------------
# PART 2 — GRU Model (configurable hidden units)
# ---------------------------------------------------------------
input_size = 2
hidden_size = 4  # Start with 4, can scale to 8, 16, 32, etc.

class GRURegressor(nn.Module):
    def __init__(self, hidden_dim):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.gru = nn.GRU(input_size, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x, h0=None):
        x = x.unsqueeze(1)
        _, h = self.gru(x, h0)
        out = self.fc(h.squeeze(0))
        return torch.sigmoid(out), h

model = GRURegressor(hidden_size)
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# ---------------------------------------------------------------
# PART 3 — State space & Monte-Carlo params
# ---------------------------------------------------------------
n_nodes = hidden_size
states = list(itertools.product([0,1], repeat=n_nodes))
num_states = len(states)

mc_inputs_per_state = 400
mc_bernoulli_samples = 4
rng = np.random.RandomState(42)

connectivity = np.ones((n_nodes, n_nodes), dtype=int)

def idx_to_state(idx):
    return np.array(list(np.binary_repr(idx, width=n_nodes))).astype(int)

# ---------------------------------------------------------------
# PART 4 — Compute Big Φ (IIT / SIA) - Only for small models
# ---------------------------------------------------------------
def compute_big_phi_and_tpm(model):
    """Only feasible for hidden_size <= 6"""
    if hidden_size > 6:
        print("⚠️  Skipping full IIT Φ computation (too large)")
        return None, None
    
    tpm_counts = np.zeros((num_states, num_states))

    with torch.no_grad():
        for i, h_state in enumerate(states):
            h0 = torch.tensor(h_state, dtype=torch.float32).view(1,1,-1)

            for _ in range(mc_inputs_per_state):
                x = rng.randint(0,2,size=(input_size,)).astype(np.float32)
                x_tensor = torch.tensor(x).view(1,1,-1)

                _, h_next = model.gru(x_tensor, h0)
                probs = torch.sigmoid(h_next).squeeze().numpy()

                for _ in range(mc_bernoulli_samples):
                    s = (rng.rand(n_nodes) < probs).astype(int)
                    j = int("".join(map(str,s)),2)
                    tpm_counts[i,j] += 1

    tpm = tpm_counts + 1e-8
    tpm /= tpm.sum(axis=1, keepdims=True)

    network = pyphi.Network(tpm, connectivity)

    phi_vals = []
    for i in range(num_states):
        try:
            subsystem = pyphi.Subsystem(network, idx_to_state(i), tuple(range(n_nodes)))
            sia = pyphi.compute.sia(subsystem)
            phi_vals.append(sia.phi)
        except:
            phi_vals.append(0.0)

    return max(phi_vals), tpm

# ---------------------------------------------------------------
# PART 5 — Φ* (Barrett & Seth) - STOCHASTIC VERSION
# ---------------------------------------------------------------
def collect_hidden_states(model, n_samples=2000):
    """Collect hidden state trajectories from the GRU"""
    states_list = []
    
    with torch.no_grad():
        h = None
        for _ in range(n_samples):
            x = rng.randint(0, 2, size=(1, input_size)).astype(np.float32)
            x_tensor = torch.tensor(x).view(1, 1, -1)
            
            _, h = model.gru(x_tensor, h)
            states_list.append(h.squeeze().numpy())
    
    return np.array(states_list)

class BarrettSethPhiStar:
    """
    TRUE Barrett & Seth (2011) Φ* for continuous Gaussian systems
    Monte-Carlo approximation of the Minimum Information Partition (MIP)
    """

    def __init__(self, n_partitions_sample=20, tau=1):
        self.n_partitions_sample = n_partitions_sample
        self.tau = tau

    def gaussian_mi(self, X, Y):
        """Gaussian mutual information I(X;Y)"""
        n = X.shape[1]
        reg = 1e-6 * np.eye(n)

        Cov_X = np.cov(X.T) + reg
        Cov_Y = np.cov(Y.T) + reg
        Cov_XY = np.cov(np.hstack([X, Y]).T) + 1e-6 * np.eye(2*n)

        sx, logdet_x = np.linalg.slogdet(Cov_X)
        sy, logdet_y = np.linalg.slogdet(Cov_Y)
        sxy, logdet_xy = np.linalg.slogdet(Cov_XY)

        if sx <= 0 or sy <= 0 or sxy <= 0:
            return 0.0

        return 0.5 * (logdet_x + logdet_y - logdet_xy) / np.log(2)

    def estimate(self, states):
        X = states[:-self.tau]
        Y = states[self.tau:]

        n_samples, n_dim = X.shape

        # ---- Whole system information ----
        I_whole = self.gaussian_mi(X, Y)

        if I_whole <= 0:
            return 0.0

        min_I_partition = np.inf
        nodes = np.arange(n_dim)

        np.random.seed(42)

        # ---- Monte-Carlo MIP search ----
        for _ in range(self.n_partitions_sample):

            k = np.random.randint(1, n_dim // 2 + 1)
            A = np.random.choice(nodes, k, replace=False)
            B = np.array(list(set(nodes) - set(A)))

            if len(B) == 0:
                continue

            I_A = self.gaussian_mi(X[:, A], Y[:, A])
            I_B = self.gaussian_mi(X[:, B], Y[:, B])

            I_partition = I_A + I_B
            min_I_partition = min(min_I_partition, I_partition)

        phi_star = I_whole - min_I_partition
        return max(phi_star, 0.0)

def compute_phi_star_barrett_stochastic(model, n_samples=2000, n_partitions=20):
    """
    Stochastic Barrett & Seth Φ* using Monte-Carlo partition sampling
    """
    states = collect_hidden_states(model, n_samples)
    estimator = BarrettSethPhiStar(n_partitions_sample=n_partitions, tau=1)
    return estimator.estimate(states)

# ---------------------------------------------------------------
# PART 6 — Decoder-based Φ* with Gaussian Approximation 
# ---------------------------------------------------------------
def compute_phi_star_decoder_gaussian(model, n_samples=2000):
    """
    Decoder-based Φ* using Gaussian approximation.
    Measures information loss when splitting the system.
    
    This scales to large models! O(n³) instead of O(2^n)
    """
    # Collect state trajectories
    states = collect_hidden_states(model, n_samples)
    n_nodes = states.shape[1]
    
    # Compute covariance matrices for Gaussian approximation
    # States at time t and t-1
    X_t = states[1:]  # Current states
    X_tm1 = states[:-1]  # Previous states
    
    # Center the data
    X_t_mean = X_t.mean(axis=0)
    X_tm1_mean = X_tm1.mean(axis=0)
    X_t_centered = X_t - X_t_mean
    X_tm1_centered = X_tm1 - X_tm1_mean
    
    # Compute covariances
    Cov_t = np.cov(X_t_centered.T) + 1e-6 * np.eye(n_nodes)
    Cov_tm1 = np.cov(X_tm1_centered.T) + 1e-6 * np.eye(n_nodes)
    Cov_joint = np.cov(np.hstack([X_tm1_centered, X_t_centered]).T)
    
    # Total mutual information I(X_t; X_{t-1})
    # For Gaussian: MI = 0.5 * log(|Cov_t| * |Cov_tm1| / |Cov_joint|)
    sign_t, logdet_t = np.linalg.slogdet(Cov_t)
    sign_tm1, logdet_tm1 = np.linalg.slogdet(Cov_tm1)
    sign_joint, logdet_joint = np.linalg.slogdet(Cov_joint)
    
    I_whole = 0.5 * (logdet_t + logdet_tm1 - logdet_joint) / np.log(2)  # Convert to bits
    
    # Find minimum information partition (MIP)
    nodes = list(range(n_nodes))
    min_phi_decoder = np.inf
    
    for k in range(1, n_nodes // 2 + 1):
        for partition in combinations(nodes, k):
            part_A = list(partition)
            part_B = list(set(nodes) - set(partition))
            
            # Compute MI for each partition independently
            I_cut = 0.0
            
            for part in [part_A, part_B]:
                if len(part) == 0:
                    continue
                
                # Extract submatrices for this partition
                Cov_t_sub = Cov_t[np.ix_(part, part)]
                Cov_tm1_sub = Cov_tm1[np.ix_(part, part)]
                
                # Joint covariance for this partition
                part_indices = part + [p + n_nodes for p in part]
                Cov_joint_sub = Cov_joint[np.ix_(part_indices, part_indices)]
                
                # MI for this partition
                sign_t_sub, logdet_t_sub = np.linalg.slogdet(Cov_t_sub)
                sign_tm1_sub, logdet_tm1_sub = np.linalg.slogdet(Cov_tm1_sub)
                sign_joint_sub, logdet_joint_sub = np.linalg.slogdet(Cov_joint_sub)
                
                I_part = 0.5 * (logdet_t_sub + logdet_tm1_sub - logdet_joint_sub) / np.log(2)
                I_cut += I_part
            
            # Φ for this partition
            phi_partition = I_whole - I_cut
            min_phi_decoder = min(min_phi_decoder, phi_partition)
    
    return max(min_phi_decoder, 0.0)

# ---------------------------------------------------------------
# PART 7 — Φ Approximation for Large Models (Tononi et al. 2016)
# ---------------------------------------------------------------
def compute_phi_approximation(model, n_samples=2000):
 
    # Collect state trajectories
    states = collect_hidden_states(model, n_samples)
    n_nodes = states.shape[1]
    
    X_t = states[1:]
    X_tm1 = states[:-1]
    
    # Center data
    X_t_centered = X_t - X_t.mean(axis=0)
    X_tm1_centered = X_tm1 - X_tm1.mean(axis=0)
    
    # Estimate interaction matrix via linear regression
    # X_t ≈ W @ X_{t-1}
    W = np.linalg.lstsq(X_tm1_centered, X_t_centered, rcond=None)[0].T
    
    # Compute covariance matrices
    Cov_t = np.cov(X_t_centered.T) + 1e-6 * np.eye(n_nodes)
    Cov_tm1 = np.cov(X_tm1_centered.T) + 1e-6 * np.eye(n_nodes)
    
    # SVD-based approximation
    U, S, Vt = np.linalg.svd(W)
    
    # Φ approximation using effective information
    # Sum of squared singular values weighted by variance
    var_tm1 = np.diag(Cov_tm1)
    phi_approx = 0.0
    
    for i in range(min(n_nodes, len(S))):
        # Effective information carried by each singular vector
        ei = S[i]**2 * var_tm1[i] if i < len(var_tm1) else S[i]**2
        phi_approx += ei
    
    # Normalize by system size for comparability
    phi_approx = phi_approx / n_nodes
    
    return phi_approx

# ---------------------------------------------------------------
# PART 8 — Accuracy
# ---------------------------------------------------------------
def compute_accuracy(model):
    with torch.no_grad():
        preds,_ = model(torch.tensor(X_test))
        preds = (preds.squeeze() > 0.5).float()
        return (preds == torch.tensor(y_test)).float().mean().item()

# ---------------------------------------------------------------
# PART 9 — Training + All Φ measures
# ---------------------------------------------------------------
EPOCHS = 10
phi_list, phi_star_barrett_stochastic_list, phi_star_decoder_list, phi_approx_list, acc_list = [], [], [], [], []

print(f"\n{'='*60}")
print(f"TRAINING GRU (hidden_size={hidden_size})")
print(f"{'='*60}\n")

for ep in range(1, EPOCHS+1):
    model.train()
    for xb,yb in train_loader:
        optimizer.zero_grad()
        pred,_ = model(xb)
        loss = criterion(pred.squeeze(), yb)
        loss.backward()
        optimizer.step()

    acc = compute_accuracy(model)
    
    # Full IIT Φ (only for small models)
    if hidden_size <= 6:
        phi, tpm = compute_big_phi_and_tpm(model)
    else:
        phi = None
    
    # Stochastic Barrett & Seth Φ* (NEW - using Monte-Carlo sampling)
    phi_star_barrett_stochastic = compute_phi_star_barrett_stochastic(model, n_samples=2000, n_partitions=20)
    
    # Scalable measures (work for any size!)
    phi_star_decoder = compute_phi_star_decoder_gaussian(model)
    phi_approx = compute_phi_approximation(model)
    
    phi_list.append(phi)
    phi_star_barrett_stochastic_list.append(phi_star_barrett_stochastic)
    phi_star_decoder_list.append(phi_star_decoder)
    phi_approx_list.append(phi_approx)
    acc_list.append(acc)

    if phi is not None:
        print(f"Epoch {ep} | Acc={acc:.3f} | Φ_IIT={phi:.6f} | Φ*_Barrett_Stoch={phi_star_barrett_stochastic:.6f} | Φ*_Decoder={phi_star_decoder:.6f} | Φ_Approx={phi_approx:.6f}")
    else:
        print(f"Epoch {ep} | Acc={acc:.3f} | Φ*_Barrett_Stoch={phi_star_barrett_stochastic:.6f} | Φ*_Decoder={phi_star_decoder:.6f} | Φ_Approx={phi_approx:.6f}")

# ---------------------------------------------------------------
# PART 10 — Plots & Correlations
# ---------------------------------------------------------------
print(f"\n{'='*60}")
print("CORRELATIONS")
print(f"{'='*60}")

if hidden_size <= 6:
    # Plot all measures
    plt.figure(figsize=(12,6))
    plt.plot(phi_list, label="Φ (IIT)", marker='o')
    plt.plot(phi_star_barrett_stochastic_list, label="Φ*", marker='s')
    plt.plot(phi_star_decoder_list, label="Φ* (Decoder)", marker='^')
    plt.plot(phi_approx_list, label="Φ (Approximation)", marker='d')
    plt.xlabel("Epoch")
    plt.ylabel("Value")
    plt.title(f"Integration Measures vs Epoch (hidden_size={hidden_size})")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Correlation matrix
    measures = {
        'Φ_IIT': phi_list,
        'Φ*_Barrett_Stoch': phi_star_barrett_stochastic_list,
        'Φ*_Decoder': phi_star_decoder_list,
        'Φ_Approx': phi_approx_list
    }
    
    print("\nPairwise correlations:")
    for m1 in measures:
        for m2 in measures:
            if m1 < m2:
                r, p = pearsonr(measures[m1], measures[m2])
                print(f"  {m1} vs {m2}: r={r:.3f}, p={p:.4f}")
    
    # Scatter plots
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    
    axes[0,0].scatter(phi_list, phi_star_barrett_stochastic_list)
    axes[0,0].set_xlabel("Φ (IIT)")
    axes[0,0].set_ylabel("Φ*")
    axes[0,0].set_title("IIT vs Φ*")
    axes[0,0].grid(True)
    
    axes[0,1].scatter(phi_list, phi_star_decoder_list)
    axes[0,1].set_xlabel("Φ (IIT)")
    axes[0,1].set_ylabel("Φ* (Decoder)")
    axes[0,1].set_title("IIT vs Decoder")
    axes[0,1].grid(True)
    
    axes[0,2].scatter(phi_list, phi_approx_list)
    axes[0,2].set_xlabel("Φ (IIT)")
    axes[0,2].set_ylabel("Φ (Approximation)")
    axes[0,2].set_title("IIT vs Approximation")
    axes[0,2].grid(True)
    
    axes[1,0].scatter(phi_star_barrett_stochastic_list, phi_star_decoder_list)
    axes[1,0].set_xlabel("Φ*")
    axes[1,0].set_ylabel("Φ* (Decoder)")
    axes[1,0].set_title("Φ* vs Decoder")
    axes[1,0].grid(True)
    
    axes[1,1].scatter(phi_approx_list, phi_star_decoder_list)
    axes[1,1].set_xlabel("Φ (Approximation)")
    axes[1,1].set_ylabel("Φ* (Decoder)")
    axes[1,1].set_title("Approximation vs Decoder")
    axes[1,1].grid(True)
    
    axes[1,2].scatter(phi_star_barrett_stochastic_list, phi_approx_list)
    axes[1,2].set_xlabel("Φ*")
    axes[1,2].set_ylabel("Φ (Approximation)")
    axes[1,2].set_title("Φ* vs Approximation")
    axes[1,2].grid(True)
    
    plt.tight_layout()
    plt.show()
    
else:
    # For large models, only plot scalable measures
    plt.figure(figsize=(8,5))
    plt.plot(phi_star_barrett_stochastic_list, label="Φ* (Barrett Stochastic)", marker='s')
    plt.plot(phi_star_decoder_list, label="Φ* (Decoder)", marker='^')
    plt.plot(phi_approx_list, label="Φ (Approximation)", marker='d')
    plt.xlabel("Epoch")
    plt.ylabel("Value")
    plt.title(f"Scalable Integration Measures (hidden_size={hidden_size})")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    print("\nPairwise correlations:")
    r1, p1 = pearsonr(phi_star_barrett_stochastic_list, phi_star_decoder_list)
    print(f"  Φ*_Barrett_Stoch vs Φ*_Decoder: r={r1:.3f}, p={p1:.4f}")
    
    r2, p2 = pearsonr(phi_approx_list, phi_star_decoder_list)
    print(f"  Φ_Approx vs Φ*_Decoder: r={r2:.3f}, p={p2:.4f}")
    
    r3, p3 = pearsonr(phi_star_barrett_stochastic_list, phi_approx_list)
    print(f"  Φ*_Barrett_Stoch vs Φ_Approx: r={r3:.3f}, p={p3:.4f}")

```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import itertools
import matplotlib.pyplot as plt
import pyphi
from itertools import combinations
from scipy.stats import pearsonr

pyphi.config.VALIDATE_SUBSYSTEM_STATES = False
pyphi.config.VALIDATE_CONDITIONAL_INDEPENDENCE = False

device = torch.device("cpu")

# ------------------- Parity Dataset -------------------
def generate_parity_data(n_samples=2000):
    X = np.random.randint(0, 2, (n_samples, 2))
    y = np.sum(X, axis=1) % 2
    return X.astype(np.float32), y.astype(np.float32)

X, y = generate_parity_data()
X_train, y_train = X[:1600], y[:1600]
X_test, y_test = X[1600:], y[1600:]

train_loader = DataLoader(
    TensorDataset(torch.tensor(X_train), torch.tensor(y_train)),
    batch_size=64,
    shuffle=True
)

# ------------------- GRU Model -------------------
input_size = 2
hidden_size = 4

class GRURegressor(nn.Module):
    def __init__(self, hidden_dim):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.gru = nn.GRU(input_size, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x, h0=None):
        x = x.unsqueeze(1)
        _, h = self.gru(x, h0)
        out = self.fc(h.squeeze(0))
        return torch.sigmoid(out), h

model = GRURegressor(hidden_size)
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# ------------------- State Space -------------------
n_nodes = hidden_size
states = list(itertools.product([0,1], repeat=n_nodes))
num_states = len(states)

mc_inputs_per_state = 400
mc_bernoulli_samples = 4
rng = np.random.RandomState(42)
connectivity = np.ones((n_nodes, n_nodes), dtype=int)

def idx_to_state(idx):
    return np.array(list(np.binary_repr(idx, width=n_nodes))).astype(int)

# ------------------- IIT Φ (small models only) -------------------
def compute_big_phi_and_tpm(model):
    if hidden_size > 6:
        return None, None
    
    tpm_counts = np.zeros((num_states, num_states))
    with torch.no_grad():
        for i, h_state in enumerate(states):
            h0 = torch.tensor(h_state, dtype=torch.float32).view(1,1,-1)
            for _ in range(mc_inputs_per_state):
                x = rng.randint(0,2,size=(input_size,)).astype(np.float32)
                x_tensor = torch.tensor(x).view(1,1,-1)
                _, h_next = model.gru(x_tensor, h0)
                probs = torch.sigmoid(h_next).squeeze().numpy()
                for _ in range(mc_bernoulli_samples):
                    s = (rng.rand(n_nodes) < probs).astype(int)
                    j = int("".join(map(str,s)),2)
                    tpm_counts[i,j] += 1

    tpm = tpm_counts + 1e-8
    tpm /= tpm.sum(axis=1, keepdims=True)

    network = pyphi.Network(tpm, connectivity)
    phi_vals = []
    for i in range(num_states):
        try:
            subsystem = pyphi.Subsystem(network, idx_to_state(i), tuple(range(n_nodes)))
            sia = pyphi.compute.sia(subsystem)
            phi_vals.append(sia.phi)
        except:
            phi_vals.append(0.0)
    return max(phi_vals), tpm

# ------------------- Hidden State Collection -------------------
def collect_hidden_states(model, n_samples=2000):
    states_list = []
    with torch.no_grad():
        h = None
        for _ in range(n_samples):
            x = rng.randint(0, 2, size=(1, input_size)).astype(np.float32)
            x_tensor = torch.tensor(x).view(1, 1, -1)
            _, h = model.gru(x_tensor, h)
            states_list.append(h.squeeze().numpy())
    return np.array(states_list)

# ------------------- Φ* Estimator -------------------
class IITAlignedPhiStar:
    def __init__(self, n_partitions_sample=100, tau=1, seed=42):
        self.n_partitions_sample = n_partitions_sample
        self.tau = tau
        self.rng = np.random.RandomState(seed)

    def gaussian_mi(self, X, Y):
        if X.shape[0] < 2 or Y.shape[0] < 2:
            return 0.0
        n_x = X.shape[1]
        n_y = Y.shape[1]
        reg_strength = max(1e-6, 0.1 / np.sqrt(X.shape[0]))
        Cov_X = np.cov(X.T) + reg_strength * np.eye(n_x)
        Cov_Y = np.cov(Y.T) + reg_strength * np.eye(n_y)
        XY_concat = np.hstack([X, Y])
        Cov_XY = np.cov(XY_concat.T) + reg_strength * np.eye(n_x + n_y)
        sx, logdet_x = np.linalg.slogdet(Cov_X)
        sy, logdet_y = np.linalg.slogdet(Cov_Y)
        sxy, logdet_xy = np.linalg.slogdet(Cov_XY)
        if sx <= 0 or sy <= 0 or sxy <= 0:
            return 0.0
        return max(0.5 * (logdet_x + logdet_y - logdet_xy) / np.log(2), 0.0)

    def estimate(self, states):
        X = states[:-self.tau]
        Y = states[self.tau:]
        n_samples, n_dim = X.shape
        if n_samples < 10:
            return 0.0

        I_whole = self.gaussian_mi(X, Y)
        if I_whole <= 0:
            return 0.0

        min_I_partition = np.inf
        nodes = np.arange(n_dim)

        if n_dim <= 4:
            for k in range(1, n_dim // 2 + 1):
                for partition in combinations(nodes, k):
                    A = np.array(partition)
                    B = np.array(list(set(nodes) - set(partition)))
                    if len(B) == 0:
                        continue
                    I_partition = self.gaussian_mi(X[:, A], Y[:, A]) + self.gaussian_mi(X[:, B], Y[:, B])
                    min_I_partition = min(min_I_partition, I_partition)
        else:
            partition_sizes = list(range(1, n_dim // 2 + 1))
            samples_per_size = max(1, self.n_partitions_sample // len(partition_sizes))
            for k in partition_sizes:
                for _ in range(samples_per_size):
                    A = self.rng.choice(nodes, k, replace=False)
                    B = np.array(list(set(nodes) - set(A)))
                    if len(B) == 0:
                        continue
                    I_partition = self.gaussian_mi(X[:, A], Y[:, A]) + self.gaussian_mi(X[:, B], Y[:, B])
                    min_I_partition = min(min_I_partition, I_partition)

        phi_star = I_whole - min_I_partition
        return max(phi_star * 0.5, 0.0)

def compute_phi_star(model, n_samples=2000, n_partitions=100):
    states = collect_hidden_states(model, n_samples)
    estimator = IITAlignedPhiStar(n_partitions_sample=n_partitions, tau=1, seed=42)
    return estimator.estimate(states)

# ------------------- Accuracy -------------------
def compute_accuracy(model):
    with torch.no_grad():
        preds,_ = model(torch.tensor(X_test))
        preds = (preds.squeeze() > 0.5).float()
        return (preds == torch.tensor(y_test)).float().mean().item()

# ------------------- Training -------------------
EPOCHS = 10
phi_iit_list = []
phi_star_list = []
acc_list = []

print(f"\n{'='*60}")
print(f"TRAINING GRU (hidden_size={hidden_size})")
print(f"Comparing IIT Φ vs Φ*")
print(f"{'='*60}\n")

for ep in range(1, EPOCHS+1):
    model.train()
    for xb,yb in train_loader:
        optimizer.zero_grad()
        pred,_ = model(xb)
        loss = criterion(pred.squeeze(), yb)
        loss.backward()
        optimizer.step()

    acc = compute_accuracy(model)
    phi_iit, _ = compute_big_phi_and_tpm(model) if hidden_size <= 6 else (None, None)
    phi_star = compute_phi_star(model, n_samples=2000, n_partitions=100)
    
    phi_iit_list.append(phi_iit)
    phi_star_list.append(phi_star)
    acc_list.append(acc)

    if phi_iit is not None:
        print(f"Epoch {ep:2d} | Acc={acc:.3f} | Φ_IIT={phi_iit:.6f} | Φ*={phi_star:.6f} | Ratio={phi_star/max(phi_iit,1e-6):.3f}")
    else:
        print(f"Epoch {ep:2d} | Acc={acc:.3f} | Φ*={phi_star:.6f}")

# ------------------- Visualization -------------------
if hidden_size <= 6 and any(p is not None for p in phi_iit_list):
    valid_indices = [i for i, p in enumerate(phi_iit_list) if p is not None]
    phi_iit_clean = [phi_iit_list[i] for i in valid_indices]
    phi_star_clean = [phi_star_list[i] for i in valid_indices]
    epochs_clean = [i+1 for i in valid_indices]

    r, p = pearsonr(phi_iit_clean, phi_star_clean)
    
    fig = plt.figure(figsize=(16, 5))
    
    ax1 = plt.subplot(1, 3, 1)
    ax1.plot(epochs_clean, phi_iit_clean, 'o-', label='IIT Φ', linewidth=2, markersize=8)
    ax1.plot(epochs_clean, phi_star_clean, 's-', label='Φ*', linewidth=2, markersize=8)
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Integration Value')
    ax1.set_title('IIT Φ vs Φ* During Training')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    ax2 = plt.subplot(1, 3, 2)
    ax2.scatter(phi_iit_clean, phi_star_clean, s=100, alpha=0.7, edgecolors='black')
    z = np.polyfit(phi_iit_clean, phi_star_clean, 1)
    p_fit = np.poly1d(z)
    x_line = np.linspace(min(phi_iit_clean), max(phi_iit_clean), 100)
    ax2.plot(x_line, p_fit(x_line), "r--", alpha=0.8, linewidth=2, label=f'Linear fit (r={r:.3f})')
    ax2.set_xlabel('IIT Φ')
    ax2.set_ylabel('Φ*')
    ax2.set_title(f'Correlation: r={r:.3f}, p={p:.4f}')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    ax3 = plt.subplot(1, 3, 3)
    residuals = np.array(phi_star_clean) - p_fit(np.array(phi_iit_clean))
    ax3.scatter(phi_iit_clean, residuals, s=100, alpha=0.7, edgecolors='black')
    ax3.axhline(y=0, color='r', linestyle='--', linewidth=2)
    ax3.set_xlabel('IIT Φ')
    ax3.set_ylabel('Residual (Φ* - Predicted)')
    ax3.set_title('Residual Analysis')
    ax3.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"\nMean IIT Φ: {np.mean(phi_iit_clean):.6f} ± {np.std(phi_iit_clean):.6f}")
    print(f"Mean Φ*:      {np.mean(phi_star_clean):.6f} ± {np.std(phi_star_clean):.6f}")
    print(f"Ratio (Φ*/Φ_IIT): {np.mean(phi_star_clean)/np.mean(phi_iit_clean):.3f}")
    print(f"Mean residual: {np.mean(np.abs(residuals)):.6f}")

else:
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, EPOCHS+1), phi_star_list, 'o-', linewidth=2, markersize=8)
    plt.xlabel('Epoch')
    plt.ylabel('Φ*')
    plt.title(f'Φ* During Training (hidden_size={hidden_size})')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```
