```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import font_manager
# -------------------------------
# CSV LOCATION 
# -------------------------------
CSV_DIR = "./CSV FIles"
SAVE_DIR = "./PNG Files(TimesNewRoman)"
os.makedirs(SAVE_DIR, exist_ok=True)

# -------------------------------
# Load CSV files
# -------------------------------
df_full = pd.read_csv(f"{CSV_DIR}/training_summary.csv")

epochs = df_full["epoch"]
accuracy = df_full["accuracy"]
phi = df_full["phi"]

print("CSV loaded successfully")
df_full.head()
```

```python
plt.figure(figsize=(12, 7))
plt.plot(epochs, phi, linewidth=2.5)
plt.xlabel("Epoch")
plt.ylabel("Φ*")
plt.title("Φ* vs Epoch")
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.savefig(f"{SAVE_DIR}/phi_vs_epoch.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(12, 7))
plt.plot(epochs, accuracy, linewidth=2.5)
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Epoch")
plt.ylim(0, 1.05)
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.savefig(f"{SAVE_DIR}/accuracy_vs_epoch.png", dpi=300)
plt.show()
```

```python
from scipy.stats import pearsonr

r, p = pearsonr(accuracy, phi)

plt.figure(figsize=(12, 7))
plt.scatter(accuracy, phi, s=100, alpha=0.8)
plt.xlabel("Accuracy")
plt.ylabel("Φ*")
plt.title("Correlation between Φ* VS Accuracy")

plt.text(
    0.05, 0.95,
    f"r = {r:.3f}\np= {p:.2e}",
    transform=plt.gca().transAxes,
    fontsize=14,
    verticalalignment="top",
    bbox=dict(boxstyle="round", alpha=0.3)
)

plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.savefig(f"{SAVE_DIR}/phi_accuracy_correlation.png", dpi=300)
plt.show()

print(f"Pearson r = {r:.3f}, p = {p:.2e}")
```
